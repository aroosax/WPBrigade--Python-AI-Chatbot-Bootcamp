```python
#importing libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
#load dataset(wine dataset)
data = load_wine(as_frame=True)
df = data.frame
```

```python
# printing first few rows of the data
df.head()
```

```python
# checking the null values in every colum
df.isnull().sum()
```

```python
df.info()
```

```python
df.describe()
```

```python
# data preprocessing
X = df.drop('target', axis=1)
Y = df['target']
```

```python
X.head()
```

```python
Y.head()
```

```python
# splitting the data
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2, random_state=42, stratify=Y)
```

```python
#model training
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train,Y_train)
#prediction
Y_pred = rf.predict(X_test)
```

```python
# Evaluation metrices
accuracy = accuracy_score(Y_test,Y_pred)
print(f'Accuracy:{accuracy:.2f}')
conf_matrix = confusion_matrix(Y_test,Y_pred)
print(f'Confusion Matrix:\n{conf_matrix}')
class_report = classification_report(Y_test,Y_pred)
print('\nClassification Report:\n',class_report)
```

```python
#feature importance
importance = rf.feature_importances_
features = X.columns
feature_importance = pd.Series(importance, index=features).sort_values(ascending=True)
print(feature_importance)
```

```python
#visualizing it
plt.figure(figsize=(10,6))
feature_importance.plot(kind='barh', title='Feature Importance')
plt.xlabel('Importance score')
plt.show()
```
