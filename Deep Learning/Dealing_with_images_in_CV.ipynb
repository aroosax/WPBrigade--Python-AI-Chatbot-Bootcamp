# show an image

```python
from google.colab.patches import cv2_imshow
```

```python
cv2_imshow(image)
```

#channels (BRG)

```python
image.shape
```

```python
#RGB -----> BLUE GREEN RED
image1 = image.copy()
image1[:,:,1]=0 # there are 3 colors BGR and 1 represents green color is not there
cv2_imshow(image1)
```

# convert into gray scale

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
gray.shape
```

#Flip an image

```python
cv2.flip(image,0)
# 0 for vertical flip
#1 for horizontal flip
#-1 for both
```

#Crop an image

```python
image
```

```python
cv2_imshow(image[1:250,1:370,:])
```

#Blur an image

```python
cv2.GaussianBlur(image,ksize=(7,7),sigmaX=0,sigmaY=0)
```

#Resize an image

```python
image
```

```python
cv2_imshow(cv2.resize(image,dsize=(300,300)))
```

#Rescale an image

```python
image.shape
```

```python
#aspect ratio same
factor = 0.7
height = int(image.shape[0]*factor)
width = int(image.shape[1]*factor)
#print(height)
cv2_imshow(cv2.resize(image,dsize=(height,width)))
```

#Circle on image

```python
image
```

```python
cv2.circle(image,center=(340,200),radius=200,color=(255,0,0),thickness=2)
cv2_imshow(image)
```

#Rectangle on image

```python
cv2.rectangle(image, pt1=(1,1),pt2=(340,200), color=(0,0,255),thickness=2)
cv2_imshow(image)
```

#Line on image

```python
cv2.line(image,pt1=(1,1),pt2=(250,250), color=(0,255,0),thickness=3)
cv2_imshow(image)
```

#Text in image

```python
cv2.putText(image,text='Find me',org=(250,250),fontFace=cv2.FONT_ITALIC,fontScale=1,color=(0,0,255),thickness=3)
cv2_imshow(image)
```
