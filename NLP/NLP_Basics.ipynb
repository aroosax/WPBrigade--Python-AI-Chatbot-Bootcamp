```python
corpus = """Hello, Welcome to Python and AI bootcamp. This is  Aroosa, i'm pleased to teach y'all.
This is where you will start to learn, grow and become expert in NLP with your continuous effort.
"""
```

```python
print(corpus)
```

```python
#tokenization
#sentence --> Paragraph
from nltk.tokenize import sent_tokenize
```

```python
import nltk
nltk.download('wordnet')
nltk.download('punkt_tab')
```

```python
documents = sent_tokenize(corpus)
type(documents)
```

```python
for sentence in documents:
  print(sentence)
```

```python
#tokenization
#paragraph/sentence to words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
  print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer = TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```

```python
#stemming and lemmatization
from nltk.stem import PorterStemmer, WordNetLemmatizer

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

words = ['studies','studying','better','children']

print('stemming')

for w in words:
  print(w, '-->',stemmer.stem(w))


print('/n Lemmatization')
for w in words:
  print(w,'-->',lemmatizer.lemmatize(w))
```
